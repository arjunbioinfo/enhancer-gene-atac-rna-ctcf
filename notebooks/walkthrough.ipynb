# Enhancer–Gene ATAC-seq + RNA-seq + CTCF Integration Walkthrough

This notebook inspects the outputs produced by the Snakemake workflow (`make ci` for `chr8` or `make run` genome-wide) comparing **K562** vs **GM12878** ENCODE GRCh38 profiles.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

RESULTS = Path('../results')
FIGURES = Path('../figures')

genes = pd.read_csv(RESULTS / 'gene_regulatory_summary.tsv', sep='\t')
stats = pd.read_csv(RESULTS / 'ctcf_enrichment_stats.tsv', sep='\t')
candidates = pd.read_csv(RESULTS / 'prioritized_candidates.tsv', sep='\t')

print(f'Genes summarized: {len(genes):,}')
print(genes['regulatory_class'].value_counts())
```

```python
# Top 15 prioritized enhancer-gene-CTCF candidates
cols = ['gene_name', 'chrom', 'rna_log2fc', 'n_distal_enhancers', 'n_ctcf_bound_enhancers', 'weighted_enhancer_score', 'regulatory_class', 'priority_score']
candidates[cols].head(15)
```

```python
# Inspect the 8q24.21 MYC / PVT1 / CCAT1 regulatory locus
candidates[candidates['gene_name'].isin(['MYC', 'PVT1', 'CCAT1'])][cols]
```

```python
# CTCF enrichment statistics (Fisher's exact test + BH FDR)
stats[['comparison', 'fg_bound', 'fg_total', 'bg_bound', 'bg_total', 'odds_ratio', 'pvalue', 'fdr_bh']]
```

```python
# Display generated summary figures
for fig_name in ['atac_rna_correlation.png', 'ctcf_enrichment.png', 'distance_to_ctcf.png', 'top_candidates_summary.png']:
    fig_path = FIGURES / fig_name
    if fig_path.exists():
        display(Image(filename=str(fig_path), width=650))
```
